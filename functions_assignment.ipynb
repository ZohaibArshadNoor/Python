## Write a function is_strong(password) that checks if the password:

* Has at least 8 characters

* Contains numbers

* Contains uppercase letters

* Contains lowercase letters

### Return True or False.

```python
def is_strong(password):
    if len(password) < 8:
        return False
    if not any(char.isupper() for char in password):
        return False   
    if not any(char.islower() for char in password):
        return False
    if not any(char.isdigit() for char in password):
        return False
    return True

is_strong("Password123")
```

## Write a function process_transaction(balance, amount, transaction_type) that supports:

* "deposit" → add amount

* "withdraw" → subtract amount

* "transfer" → subtract amount + 1% transfer fee

### Return the updated balance.
### If amount > balance (for withdraw/transfer), return "Transaction Failed".

```python
def process_transaction(balance, amount, transaction_type):
    if transaction_type == "deposit":
        return balance + amount
    elif transaction_type == "withdraw":
        if amount > balance:
            return "Transaction Failed"
        return balance - amount
    elif transaction_type == "transfer":
        fee = amount * 0.01
        total = amount + fee
        if total > balance:
            return "Transaction Failed"
        return balance - total
    else:
        return "Invalid transaction type"
 
print(process_transaction(1000, 200, "deposit"))   # 1200
print(process_transaction(1000, 200, "withdraw"))  # 800
print(process_transaction(1000, 200, "transfer"))  # 1000 - 202 = 798
print(process_transaction(1000, 1200, "withdraw")) # Transaction Failed
```

## Write a function analyze_results(marks_list) 
### that returns:
    Highest marks

    Lowest marks
    
    Average
    
    Pass rate percentage
    Passing marks = 50
    
### Return the data in a dictionary.

```python
def analyze_results(marks_list):
    if not marks_list:
        return {}
    highest = max(marks_list)
    lowest = min(marks_list)
    avg = sum(marks_list) / len(marks_list)
    passes = sum(1 for m in marks_list if m >= 50)
    pass_rate = (passes / len(marks_list)) * 100
    return {
        "highest": highest,
        "lowest": lowest,
        "average": avg,
        "pass_rate": pass_rate
    }

analyze_results([70, 45, 90, 55])
```

## Airline Seat Booking With Constraints

### Write a function book_seat(seats, seat_requested):

### Constraints:

    If seat_requested not in seats → return "Invalid Seat"
    
    If seat is "booked" → return "Seat Already Booked"
    
    Otherwise book the seat and return updated seats dictionary

### Example seats dictionary:
{
  "1A": "free",
  "1B": "booked",
  "2A": "free"
}

```python
def book_seat(seats, seat_requested):
    if seat_requested not in seats:
        return "Invalid Seat"
    if seats[seat_requested] == "booked":
        return "Seat Already Booked"
    seats[seat_requested] = "booked"
    return seats

seats = {"1A": "free", "1B": "booked", "2A": "free"}
book_seat(seats, "2A")
```

## Write a function calculate_tax(salary) using slabs:

* 0–50,000 → 0%

* 50,001–100,000 → 5%

* 100,001–200,000 → 12%

* Above 200,000 → 20%

### Return total tax payable.

```python
def calculate_tax(salary):
    tax = 0
    if salary > 200000:
        tax += (salary - 200000) * 0.20
        tax += 100000 * 0.12
        tax += 50000 * 0.05
    elif salary > 100000:
        tax += (salary - 100000) * 0.12
        tax += 50000 * 0.05
    elif salary > 50000:
        tax += (salary - 50000) * 0.05
    return tax

calculate_tax(250000)
```

## Create a Python program that:

### Defines a function called generate_salary(employee)

####    The employee parameter will be a dictionary containing:

        "name" → employee name
        
        "basic" → basic salary
        
        "overtime_hours" → extra hours worked
        
        "performance" → string ("excellent", "good", "average", "poor")

### The function must:

#####    ✔ 1. Calculate Gross Salary
    
    Gross = basic salary
    
    overtime pay
    
    performance bonus
    
#####    ✔ 2. Overtime Pay Rules
    
    1 overtime hour = 300 PKR
    
####    ✔ 3. Performance Bonus Rules
    
    excellent → 20% of basic
    
    good → 10%
    
    average → 5%
    
    poor → 0%
    
#####    ✔ 4. Apply Deductions
    
    Income tax = 12% of gross
    
    Medical insurance = 1000 PKR
    
#####    ✔ 5. Return & print a full salary slip with this format:
        Employee: Ali
        Basic Salary: 50000
        Overtime Pay: 6000
        Performance Bonus: 10000
        Gross Salary: 66000
        Income Tax (12%): 7920
        Medical Insurance: 1000
        Net Salary: 57080

### 📝 Example Input:
    employee = {
        "name": "Ali",
        "basic": 50000,
        "overtime_hours": 20,
        "performance": "excellent"
    }

### 📝 Example Output:
    Employee: Ali
    Basic Salary: 50000
    Overtime Pay: 6000
    Performance Bonus: 10000
    Gross Salary: 66000
    Income Tax (12%): 7920
    Medical Insurance: 1000
    Net Salary: 57080

```python
def generate_salary(employee):
    basic = employee.get("basic",0)
    overtime = employee.get("overtime_hours",0) * 300
    perf = employee.get("performance","poor")
    bonus_pct = {"excellent":0.2,"good":0.10,"average":0.05,"poor":0}.get(perf,0)
    bonus = basic * bonus_pct
    gross = basic + overtime + bonus
    income_tax = gross * 0.12
    medical = 1000
    net = gross - income_tax - medical
    print(f"Employee: {employee['name']}")
    print(f"Basic Salary: {basic}")
    print(f"Overtime Pay: {overtime}")
    print(f"Performance Bonus: {bonus}")
    print(f"Gross Salary: {gross}")
    print(f"Income Tax (12%): {income_tax}")
    print(f"Medical Insurance: {medical}")
    print(f"Net Salary: {net}")
    return net

employee = {
    "name": "Ali",
    "basic": 50000,
    "overtime_hours": 20,
    "performance": "excellent"
}
generate_salary(employee)
```

### The function should:

    Calculate lesson completion %
    
    Calculate average quiz score

    ≥ 80% completion → "On Track"

    student = {
        "name": "Ubaid",
        "lessons_completed": 8,
        "total_lessons": 10,

    }

    Lesson Completion: 80%

```python
def course_progress(student):
    lessons = student.get("lessons_completed",0)
    total = student.get("total_lessons",0)
    quizzes = student.get("quizzes",[])
    lesson_pct = (lessons / total) * 100 if total else 0
    avg_quiz = sum(quizzes) / len(quizzes) if quizzes else 0
    if lesson_pct >= 80:
        status = "On Track"
    elif lesson_pct >= 50:
        status = "Needs Improvement"
    else:
        status = "Behind"
    print(f"Student: {student['name']}")
    print(f"Lesson Completion: {lesson_pct}%")
    print(f"Average Quiz Score: {avg_quiz}")
    print(f"Progress Status: {status}")
    return {"lesson_pct": lesson_pct, "avg_quiz": avg_quiz, "status": status}

student = {
    "name": "Ubaid",
    "lessons_completed": 8,
    "total_lessons": 10,
    "quizzes": [80, 75, 90]
}
course_progress(student)
```

## An online store calculates final prices dynamically:

### Function calculate_discount(price, category) → category-based discounts:

    Electronics > $1000 → 15%
    
    Clothing > $500 → 10%
    
    Groceries → 5%
    
    Function apply_coupon(price, coupon_code) → nested function applies coupon:
    
    SAVE10 → $10 off
    
    SAVE20 → $20 off if price > $200
    
    Function final_price(price, category, coupon) → combines discount + coupon + tax (8%)

### 📥 Input Example:
    final_price(1200, "Electronics", "SAVE10")

### 📝 Example Output:
    Price after discount: 1020
    Price after coupon: 1010
    Final Price with Tax: 1090.8

```python
def calculate_discount(price, category):
    if category == "Electronics" and price > 1000:
        return price * 0.15
    if category == "Clothing" and price > 500:
        return price * 0.10
    if category == "Groceries":
        return price * 0.05
    return 0

def apply_coupon(price, coupon_code):
    if coupon_code == "SAVE10":
        return price - 10
    if coupon_code == "SAVE20" and price > 200:
        return price - 20
    return price

def final_price(price, category, coupon):
    disc = calculate_discount(price, category)
    after_disc = price - disc
    after_coupon = apply_coupon(after_disc, coupon)
    tax = after_coupon * 0.08
    final = after_coupon + tax
    print(f"Price after discount: {after_disc}")
    print(f"Price after coupon: {after_coupon}")
    print(f"Final Price with Tax: {final}")
    return final

final_price(1200, "Electronics", "SAVE10")
```

## A hotel calculates room charges:

### Function room_charges(nights, room_type) → rates:

    Standard = $100/night
    
    Deluxe = $200/night
    
    Suite = $400/night
    
    Function additional_services(services) → Breakfast=$20, Spa=$50, Gym=$10
    
    Function apply_membership_discount(amount, membership_years) → 5% discount if membership > 2 years, 10% if >5 years
    
    Function total_bill(nights, room_type, services, membership_years) → prints detailed bill

### 📥 Input Example:
    total_bill(4, "Deluxe", ["Breakfast", "Spa"], 6)

### 📝 Example Output:
Room Charges: 800
Additional Services: 70
Membership Discount: 87
Total Amount: 783

```python
def room_charges(nights, room_type):
    rates = {"Standard":100, "Deluxe":200, "Suite":400}
    return rates.get(room_type, 0) * nights

def additional_services(services):
    prices = {"Breakfast":20, "Spa":50, "Gym":10}
    return sum(prices.get(s, 0) for s in services)

def apply_membership_discount(amount, membership_years):
    if membership_years > 5:
        return amount * 0.10
    if membership_years > 2:
        return amount * 0.05
    return 0

def total_bill(nights, room_type, services, membership_years):
    rc = room_charges(nights, room_type)
    add = additional_services(services)
    discount = apply_membership_discount(rc + add, membership_years)
    total = rc + add - discount
    print(f"Room Charges: {rc}")
    print(f"Additional Services: {add}")
    print(f"Membership Discount: {discount}")
    print(f"Total Amount: {total}")
    return total

total_bill(4, "Deluxe", ["Breakfast", "Spa"], 6)
```

    Function final_price(price, category, coupon) → nested calls calculate total discounts + final price

### 📥 Input Example:
    final_price(500, "Electronics", "SAVE10")

### 📝 Example Output:
    Price after base discount: 475
    Price after category discount: 427.5
    Price after coupon: 417.5
    Final Price: 451.3 (with 8% tax)

```python
def base_discount(price):
    return price * 0.05

def category_discount(price, category):
    if category == "Electronics":
        return price * 0.10
    if category == "Clothing":
        return price * 0.08
    if category == "Grocery":
        return price * 0.03
    return 0

def coupon_discount(price, coupon):
    if coupon == "SAVE10":
        return 10
    return 0

def final_price(price, category, coupon):
    p1 = price - base_discount(price)
    p2 = p1 - category_discount(p1, category)
    p3 = p2 - coupon_discount(p2, coupon)
    final = p3 + p3 * 0.08
    print(f"Price after base discount: {p1}")
    print(f"Price after category discount: {p2}")
    print(f"Price after coupon: {p3}")
    print(f"Final Price: {final}")
    return final

final_price(500, "Electronics", "SAVE10")
```
