## Student Marks Entry

### Goal:
    Record student marks and skip invalid entries.

### Steps:

    Create an empty list marks = [].
    
    Ask user to enter marks for 5 students.
    
    If marks < 0 or > 100 → print “Invalid marks” and continue.
    
    If user enters -1 → stop taking input (break).

### Example:

    Enter marks: 90

## Traffic Light System

### Goal:
    Simulate traffic light behavior.

### Task:

    Ask user to input current light color (red, yellow, or green).
    
    Use nested conditions to print what the driver should do.

### Example:

    Enter light color: red
    Stop! Wait for green light.

```python
while True:
    light_status = input("Is the light color? (green/yellow/red): ")
    if light_status.lower() == "green":
        print("Go.")
        break
    elif light_status.lower() == "yellow":
        print("Slow down.")
    elif light_status.lower() == "red":
        print("Stop.")
    else:
        print("Invalid input. Please enter green, yellow, or red.")
```

## 🏨 8. Hotel Room Booking

### Goal:
    Manage simple room booking.

### Task:

    Create a list of rooms [101, 102, 103, 104, 105]
    
    Ask user to choose a room.
    
    If room already booked, print “Room not available.”
    
    Continue until all rooms are booked.

### Example:

    Available rooms: [101, 102, 103, 104, 105]
    Choose room: 103
    Room 103 booked successfully!

```python
room_list = [101, 102, 103, 104, 105]
while True:
    if len(room_list) == 0:        
        print("No rooms available.")
        break
    print(f"Available rooms: {room_list}")
    room_number = int(input("Enter a room number to book : "))
    if room_number in room_list:
        print(f"Room {room_number} booked successfully.")
        room_list.remove(room_number)
    else:
        print(f"Room {room_number} is not available.")
```

## Shop Sales Tracker
### Goal:

    Track daily sales of 3 products over 5 days and calculate the total revenue.

### Steps:

    Outer loop → iterate over 5 days.
    
    Inner loop → take sales amount for 3 products each day.
    
    Store daily total in a list.
    
    If any product sale = 0 → continue (skip that product).
    
    Print all daily totals and grand total.

### Example:
    Day 1:
    Product 1: 200
    Product 2: 150
    Product 3: 0 → skipped
    Day 2:
    Product 1: 300
    Product 2: 250
    Product 3: 400
    → Daily Totals = [350, 950], Grand Total = 1300

```python
total_revenue = []
sales = []
for i in range(5):
    print(f"Day {i+1}: ")
    for j in range(3):
        sale = int(input(f"Sales of Product {j+1}: "))
        print(f"Product {j+1}: {sale}")
        if sale == 0:
            print("No sales for this product.")
            continue
        sales.append(sale)
    total_revenue.append(sum(sales))
    sales.clear()
print(f"Total sales for each day: {total_revenue}")
print(f"Total Revenue: {sum(total_revenue)}, Average Revenue per day: {sum(total_revenue)/len(total_revenue) if total_revenue else 0}")
    
```

## Invoice Tax Calculator
### Goal:

Take multiple items in multiple invoices and calculate tax per invoice.

### Steps:

Outer loop → for 3 invoices.

Inner loop → 4 items each.

Input item price, add to list.

Calculate total + 10% tax.

Print invoice total.

### Example:
    Invoice 1:
    Items: [100, 200, 300, 400]
    Invoice 2:
    Items: [250, 150, 350, 250]
    → Totals: [1100, 1100]

```python
totals = []

for invoice_num in range(1, 4):
    print(f"\nInvoice {invoice_num}")
    
    items = []
    
    for item_num in range(1, 5):
        price = float(input(f"Enter price for item {item_num}: "))
        items.append(price)
    
    subtotal = sum(items)
    tax = subtotal * 0.10
    total = subtotal + tax
    
    totals.append(total)
    
    print(f"Subtotal: {subtotal}")
    print(f"Tax (10%): {tax}")
    print(f"Invoice Total: {total}")

print("\nAll Invoice Totals:", totals)
```

## Student Result Sheet
### Goal:

    Generate report cards for each student in a class.

### Steps:

    Outer loop → 5 students.
    
    Inner loop → 4 subjects.

    If marks < 33 → print “Failed in Subject”.

    Calculate total and grade per student.

### Example:
    Student 1: [80, 90, 70, 85] → Grade: A  
    Student 2: [50, 32, 45, 40] → Failed in Subject 2

```python

for i in range(5):
    fail_subject = 0
    marks = []
    for j in range(4):
        mark = int(input(f'Subject {j+1}: '))
        marks.append(mark)
        if mark < 33:
            fail_subject = j+1
    if fail_subject > 0:
        print(f'Student {i+1}: {marks} Failed in Subjects {fail_subject}.')
    else:
        if sum(marks)/400*100 >= 85:
            print(f'Student {i+1}: {marks} -> Grade A.')
        elif sum(marks)/400*100 >= 70:
            print(f'Student {i+1}: {marks} -> Grade B.')
        elif sum(marks)/400*100 >= 55:
            print(f'Student {i+1}: {marks} -> Grade C.')
        else:
            print(f'Student {i+1}: {marks} -> Grade D.')
        
```
