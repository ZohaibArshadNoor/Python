1. Create a class 'BankAccount' with attributes account holder name, account number, and
balance. Add methods for deposit and withdrawal. Implement logic to prevent withdrawal if balance
is insufficient.

```python
class BankAccount:
    def __init__(self, accountholder, account_number, balance=0):
        self.accountholder = accountholder
        self.account_number = account_number
        self.balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print(f"Deposited ${amount}. New balance: ${self.balance}")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        if amount > 0:
            if self.balance >= amount:
                self.balance -= amount
                print(f"Withdrew ${amount}. New balance: ${self.balance}")
            else:
                print("Insufficient funds.")
        else:
            print("Withdrawal amount must be positive.")

    def get_balance(self):
        return self.balance
    
emp = BankAccount("John Doe", "123456789", 1000)
emp.deposit(500)
emp.withdraw(200)
print(f"Current balance: ${emp.get_balance()}")
    
```

2. Design a class 'Student' with attributes name and marks. Create a method to calculate grade
based on marks using conditions (A, B, C, Fail).

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
    
    def calculate_grade(self):
        if self.marks >= 90:
            return 'A'
        elif self.marks >= 80:
            return 'B'
        elif self.marks >= 70:
            return 'C'
        else:
            return 'Fail'
    
    def display_info(self):
        grade = self.calculate_grade()
        print(f"Student: {self.name}, Marks: {self.marks}, Grade: {grade}")

student1 = Student("Ali", 85)
student1.display_info()

student2 = Student("Babar", 92)
student2.display_info()

student3 = Student("Charles", 65)
student3.display_info()
```

3. Create a class 'ElectricityBill' for a KE-electric consumer. Include units consumed and calculate
bill based on slab rates (e.g., different price per unit for different ranges).

```python
class ElectricityBill:
    def __init__(self, consumer_name, units_consumed):
        self.consumer_name = consumer_name
        self.units_consumed = units_consumed
    
    def calculate_bill(self):
        bill = 0
        if self.units_consumed <= 100:
            bill = self.units_consumed * 50
        elif self.units_consumed <= 300:
            bill = 100 * 50 + (self.units_consumed - 100) * 75
        else:
            bill = 100 * 50 + 200 * 75 + (self.units_consumed - 300) * 100
        return bill
    
    def display_bill(self):
        bill_amount = self.calculate_bill()
        print(f"Consumer: {self.consumer_name}")
        print(f"Units Consumed: {self.units_consumed}")
        print(f"Total Bill: Rs.{bill_amount:.2f}")

bill1 = ElectricityBill("Ahmed Khan", 150)
bill1.display_bill()

bill2 = ElectricityBill("Sara Ali", 350)
bill2.display_bill()
```

4. Design a class 'Vehicle' and create subclasses 'Car' and 'Bike'. Override a method

```python
class Vehicle:
    def __init__(self, name, brand):
        self.name = name
        self.brand = brand
    
    def start_engine(self):
        print("Engine starting...")

class Car(Vehicle):
    def start_engine(self):
        print(f"{self.brand} {self.name} car engine started with a smooth roar.")

class Bike(Vehicle):
    def start_engine(self):
        print(f"{self.brand} {self.name} bike engine started with a loud buzz.")

car = Car("Civic", "Honda")
car.start_engine()

bike = Bike("CB500", "Honda")
bike.start_engine()

print("\n")

vehicles = [car, bike]
for vehicle in vehicles:
    vehicle.start_engine()
```

```python
class Employee:
    def __init__(self, name, basic_salary):
        self.name = name
        self.basic_salary = basic_salary
    
    def calculate_total_salary(self):
        bonus = self.basic_salary * 0.10
        tax = (self.basic_salary + bonus) * 0.05
        total_salary = self.basic_salary + bonus - tax
        return total_salary
    
    def display_salary_details(self):
        bonus = self.basic_salary * 0.10
        tax = (self.basic_salary + bonus) * 0.05
        total_salary = self.calculate_total_salary()
        print(f"Employee: {self.name}")
        print(f"Basic Salary: Rs.{self.basic_salary:.2f}")
        print(f"Bonus (10%): Rs.{bonus:.2f}")
        print(f"Tax (5%): Rs.{tax:.2f}")
        print(f"Total Salary: Rs.{total_salary:.2f}")

emp1 = Employee("Fatima", 50000)
emp1.display_salary_details()

print("\n")

emp2 = Employee("Hassan", 75000)
emp2.display_salary_details()
```

6. Design an abstract class 'Shape' with a method 'area()'. Create subclasses 'Circle' and
'Rectangle' and implement area calculation logic.

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def Area(self):
        pass


class Circle(Shape):
    def __init__(self, Radius):
        self.Radius = Radius

    def Area(self):
        return 3.14 * self.Radius * self.Radius


class Rectangle(Shape):
    def __init__(self, Length, Width):
        self.Length = Length
        self.Width = Width

    def Area(self):
        return self.Length * self.Width


c = Circle(5)
r = Rectangle(4, 6)

print('Circle Area:', c.Area())
print('Rectangle Area:', r.Area())
```

```python
class HospitalPatient:
    def __init__(self, Name, Age, Disease):
        self.Name = Name
        self.Age = Age
        self.Disease = Disease

    def CheckPriority(self):
        serious = ["cancer", "heart attack", "stroke"]

        if self.Age > 60 or self.Disease.lower() in serious:
            return "Emergency"
        return "Normal"


p1 = HospitalPatient("Ali", 65, "flu")
p2 = HospitalPatient("Ahmed", 30, "cancer")

print('Patient 1 Priority:', p1.CheckPriority())
print('Patient 2 Priority:', p2.CheckPriority())
```

10. Design a class 'Library' where users can borrow and return books. Add logic to check
availability before issuing a book.

```python
class Book:
    def __init__(self, id, title, author, copies):
        self.id = id
        self.title = title
        self.author = author
        self.copies = copies

    def borrow(self):
        if self.copies > 0:
            self.copies -= 1
            return True
        return False

    def give_back(self):
        self.copies += 1


class Library:
    def __init__(self):
        self.books = {}
        self.record = {}

    def add(self, id, title, author, copies):
        self.books[id] = Book(id, title, author, copies)
        print(title, "added")

    def borrow(self, user, id):
        if id not in self.books:
            print("no book")
            return

        book = self.books[id]

        if not book.borrow():
            print("not available")
            return

        if user not in self.record:
            self.record[user] = []

        self.record[user].append(id)
        print(user, "took", book.title)

    def give_back(self, user, id):
        if user not in self.record or id not in self.record[user]:
            print("wrong return")
            return

        book = self.books[id]
        book.give_back()
        self.record[user].remove(id)
        print(user, "returned", book.title)

    def check(self, id):
        if id in self.books:
            book = self.books[id]
            print(book.title, "-", book.copies, "left")
        else:
            print("no book")


lib = Library()

lib.add(1, "python", "guido", 3)
lib.add(2, "data", "john", 2)
lib.add(3, "web", "jane", 1)

lib.check(1)

lib.borrow("ali", 1)
lib.borrow("ahmed", 1)

lib.check(1)

lib.give_back("ali", 1)

lib.check(1)

lib.borrow("usman", 3)
lib.borrow("bilal", 3)
```

```python
from abc import ABC, abstractmethod
from datetime import datetime
import json
import uuid

class Flight(ABC):
    def __init__(self, flight_number, base_price):
        self.flight_number = flight_number
        self._base_price = base_price
        self.available_seats = 100
        self.bookings = []
    
    @abstractmethod
    def get_price(self):
        pass
    
    def book_seat(self, passenger, category='adult'):
        if self.available_seats <= 0:
            print("No seats available!")
            return False
        
        price = self.get_price()
        
        if category == 'child':
            price *= 0.5
        elif category == 'senior':
            price *= 0.6
        
        self.available_seats -= 1
        booking_id = str(uuid.uuid4())[:8]
        
        booking = {
            'booking_id': booking_id,
            'passenger': passenger,
            'flight': self.flight_number,
            'price': price,
            'category': category
        }
        self.bookings.append(booking)
        return booking

class Economy(Flight):
    def get_price(self):
        return self._base_price

class Business(Flight):
    def get_price(self):
        return self._base_price * 1.5

class FirstClass(Flight):
    def get_price(self):
        return self._base_price * 2.5

class Passenger:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def get_category(self):
        if self.age < 12:
            return 'child'
        elif self.age >= 60:
            return 'senior'
        return 'adult'

class PaymentProcessor:
    @staticmethod
    def process_payment(booking, payment_method='full'):
        price = booking['price']
        tax = price * 0.12
        total = price + tax
        
        if payment_method == 'installment':
            installments = total / 3
            return {'total': total, 'tax': tax, 'installment': installments}
        
        return {'total': total, 'tax': tax}

class AirlineReservationSystem:
    def __init__(self):
        self.flights = {}
    
    def create_flight(self, flight_class, flight_number, base_price):
        self.flights[flight_number] = flight_class(flight_number, base_price)
    
    def book_flight(self, flight_number, passenger_name, age, payment_method='full'):
        flight = self.flights.get(flight_number)
        passenger = Passenger(passenger_name, age)
        category = passenger.get_category()
        
        booking = flight.book_seat(passenger_name, category)
        if not booking:
            return None
        
        payment = PaymentProcessor.process_payment(booking, payment_method)
        booking['payment'] = payment
        booking['date'] = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
        
        self.save_booking(booking)
        self.print_ticket(booking)
        return booking
    
    def print_ticket(self, booking):
        print(f"\n{'='*50}")
        print(f"BOOKING CONFIRMATION")
        print(f"{'='*50}")
        print(f"Booking ID: {booking['booking_id']}")
        print(f"Passenger: {booking['passenger']} ({booking['category']})")
        print(f"Flight: {booking['flight']}")
        print(f"Base Price: Rs.{booking['price']:.2f}")
        print(f"Tax (12%): Rs.{booking['payment']['tax']:.2f}")
        print(f"Total: Rs.{booking['payment']['total']:.2f}")
        print(f"Date: {booking['date']}")
        print(f"{'='*50}\n")
    
    def save_booking(self, booking):
        with open('airline_bookings.json', 'a') as f:
            json.dump(booking, f)
            f.write('\n')

system = AirlineReservationSystem()
system.create_flight(Economy, 'EK101', 5000)
system.create_flight(Business, 'EK102', 8000)
system.create_flight(FirstClass, 'EK103', 15000)

system.book_flight('EK101', 'Ali Ahmed', 35, 'full')
system.book_flight('EK101', 'Fatima Khan', 8, 'installment')
system.book_flight('EK102', 'Hassan Ali', 65, 'full')
```

```python
from abc import ABC, abstractmethod
import json
import uuid
from datetime import datetime

class FoodItem(ABC):
    def __init__(self, name, price, quantity_available):
        self.name = name
        self._price = price
        self._quantity_available = quantity_available
    
    @property
    def price(self):
        return self._price
    
    @property
    def quantity_available(self):
        return self._quantity_available
    
    @abstractmethod
    def get_description(self):
        pass
    
    def is_available(self, quantity):
        return self._quantity_available >= quantity
    
    def reduce_quantity(self, quantity):
        if self.is_available(quantity):
            self._quantity_available -= quantity
            return True
        return False

class Pizza(FoodItem):
    def __init__(self, name, price, quantity_available, size):
        super().__init__(name, price, quantity_available)
        self.size = size
    
    def get_description(self):
        return f"Pizza: {self.name}, Size: {self.size}, Price: Rs.{self.price}"

class Burger(FoodItem):
    def __init__(self, name, price, quantity_available, type_):
        super().__init__(name, price, quantity_available)
        self.type_ = type_
    
    def get_description(self):
        return f"Burger: {self.name}, Type: {self.type_}, Price: Rs.{self.price}"

class Drink(FoodItem):
    def __init__(self, name, price, quantity_available, volume):
        super().__init__(name, price, quantity_available)
        self.volume = volume
    
    def get_description(self):
        return f"Drink: {self.name}, Volume: {self.volume}, Price: Rs.{self.price}"

class Customer:
    def __init__(self, name):
        self.name = name
        self.cart = []
    
    def add_to_cart(self, item, quantity):
        if item.is_available(quantity):
            self.cart.append({'item': item, 'quantity': quantity})
            print(f"Added {quantity} x {item.name} to cart.")
        else:
            print(f"Sorry, {item.name} is not available in sufficient quantity.")
    
    def calculate_total(self):
        total = sum(item['item'].price * item['quantity'] for item in self.cart)
        if total > 2000:
            total *= 0.9  # 10% discount
        tax = total * 0.15  # 15% tax
        return total + tax, total, tax
    
    def checkout(self, payment_method):
        if not self.cart:
            print("Cart is empty!")
            return None
        
        total_with_tax, subtotal, tax = self.calculate_total()
        order_id = str(uuid.uuid4())[:8]
        
        order = {
            'order_id': order_id,
            'customer': self.name,
            'items': [{'name': item['item'].name, 'quantity': item['quantity'], 'price': item['item'].price} for item in self.cart],
            'subtotal': subtotal,
            'discount': subtotal * 0.1 if subtotal > 2000 else 0,
            'tax': tax,
            'total': total_with_tax,
            'payment_method': payment_method,
            'date': datetime.now().strftime("%Y-%m-%d %H:%M:%S")
        }
        
        # Reduce quantities
        for item in self.cart:
            item['item'].reduce_quantity(item['quantity'])
        
        self.save_order(order)
        self.print_invoice(order)
        self.cart = []  # Clear cart
        return order
    
    def print_invoice(self, order):
        print(f"\n{'='*50}")
        print(f"INVOICE")
        print(f"{'='*50}")
        print(f"Order ID: {order['order_id']}")
        print(f"Customer: {order['customer']}")
        print(f"Items:")
        for item in order['items']:
            print(f"  - {item['name']} x{item['quantity']} @ Rs.{item['price']} each")
        print(f"Subtotal: Rs.{order['subtotal']:.2f}")
        print(f"Discount: Rs.{order['discount']:.2f}")
        print(f"Tax (15%): Rs.{order['tax']:.2f}")
        print(f"Total: Rs.{order['total']:.2f}")
        print(f"Payment Method: {order['payment_method']}")
        print(f"Date: {order['date']}")
        print(f"{'='*50}\n")
    
    def save_order(self, order):
        with open('food_orders.txt', 'a') as f:
            f.write(json.dumps(order) + '\n')

# Example usage
pizza = Pizza("Margherita", 500, 10, "Large")
burger = Burger("Cheese Burger", 300, 5, "Veg")
drink = Drink("Coke", 100, 20, "500ml")

customer = Customer("Ahmed")
customer.add_to_cart(pizza, 2)
customer.add_to_cart(burger, 1)
customer.add_to_cart(drink, 3)

customer.checkout("Card")
```
